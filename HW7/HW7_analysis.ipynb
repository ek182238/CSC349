2.) How did the Elbow Method and Silhouette Analysis in Task 2 reflect the natural structure of the Wine dataset? Did both evaluation metrics align with the true number of wine cultivars (k=3)? Explain any discrepancies observed.

**The Elbow Method measures inertia (distances within the clusters), and in its graph, the inertia decreased as the number of clusters increased. The elbow point in the graph is the inflection point. This reflects the natural structure of the wine dataset because it looked like the elbow point was around k=3, because it stopped drastically decreasing at that point. This aligns with the true number of wine cultivars (k=3). Silhouette scores measure separation, and the Silhouette analysis reflected the natural structure of the dataset because k=3 clusters had the highest silhouette score (0.28). k=2 and k=4 had the second-highest silhouette scores (0.26). So the silhouette score evaluation metrics aligned with the true number of wine cultivars because k=3 had the most separation between the clusters.**

3.) Compare the clustering behavior and cluster assignments of K-Means against DBSCAN and HDBSCAN on the 13-dimensional Wine feature space. Why did DBSCAN and HDBSCAN classify specific wine samples as noise, while K-Means assigned every sample to a cluster?

**K-Means assigned every sample to a cluster because it assigns every wine sample to the closest centroid. For example, according to the heatmap, Wine 3 (the one in row 50 in the dataset) had a distance of 3.39 from the closest centroid, which isn't as close as the other wine samples were to their nearest centroids, but it was assigned to that one anyway because the its distances to the other ones were larger. DBSCAN and HDBSCAN classified specific wine samples as noise because DBSCAN uses the epsilon distance (for example, 2.4), and HDBSCAN uses a hierarchial clustering algorithm. In DBSCAN, the data points that didn't fall into the eps=2.4 search radius were marked as noise, and in HDBSCAN, data that doesn't belong stably to a cluster is marked as noise.**

4.) Contrast the operational mechanisms of standard DBSCAN and HDBSCAN. How does HDBSCAN's hierarchical, stability-based approach overcome DBSCAN's vulnerability to a single global eps parameter when clustering chemical datasets with varying local densities?

**DBSCAN looks for at least min_samples amount of samples nearby within a specified distance (eps). If eps is too large, it might cluster lots of different groups into one cluster, but if eps is too small, lots of data points will be classified as noise. When I was testing different DBSCAN hyperparameters, I found that 2.4 eps and 4 min_samples worked best, and I found that smaller eps had more noise and lower min_samples had more clusters (with inaccurate clusters). It identified 33 noise points but only 2 clusters. HDBSCAN's approach overcomes DBSCAN's eps parameter vulnerabilities because it uses different epsilon values instead of one global epsilon value. This way, it can work with clusters of different densities and adjust its eps value based on that. In this project, HDBSCAN identified 58 noise points and 3 clusters.**
