2.) Contrast how hard voting aggregates predictions versus soft voting. Why do well-calibrated probability estimates matter when implementing soft voting across heterogeneous models?

**Hard voting aggregates predictions by using the class selected by a majority vote, and soft voting aggregates predictions based on probability (class with the highest average). Well-calibrated probability estimates matter when implementing soft voting across heterogeneous models because soft voting takes the average of each model's probability, so if a model with a really high probability isn't calibrated correctly, the prediction from soft voting may be inaccurate.**

3.) What is the role of out-of-bag evaluation in Bagging classifiers, and why does an OOB score serve as an efficient alternative to traditional cross-validation splits during ensemble training?

**Out-of-bag evaluation allows the samples that were left out of bootstrap partitions to be used for validation and testing. An OOB score serves as an efficient alternative to traditional cross-validation splits during ensemble training because you don't need to make separate validation sets or perform extra calculations.**

4.) How does introducing random feature subsetting in Random Forests differ from standard Bagging, and how does this feature diversity impact model variance and generalization?

**Introducing random feature subsetting in Random Forests differs from standard Bagging because it decorrelates the model from the 1 or 2 strong features that would be selected when using standard Bagging. Using Random Forests, a random subset of features are selected at each split. This gives Random Forests more feature diversity and reduces ensemble variance.**

5.) Discuss the mechanics of sequential sample weighting in AdaBoost. Based on your training versus test accuracy results, did AdaBoost exhibit signs of overfitting, and how can hyperparameter adjustments mitigate this?

**Sequential sample weighting in AdaBoost works by correcting errors in classification sequentially. It trains weak learners, increases the weights of misclassified samples, and then trains the next weak learner using the updated weights. Based on my results, AdaBoost did not exhibit signs of overfitting because the training set had an accuracy score of 0.7883, and the testing set had an accuracy score of 0.7857, which are very similar.
