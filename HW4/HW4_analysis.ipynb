3. Contrast how a Support Vector Classifier (using an RBF kernel) constructs decision boundaries versus how a Decision Tree recursively partitions feature space. What are the interpretability tradeoffs between the two models?

**A SVC (using an RBF kernel) constructs decision boundaries by calculating the distance between data points. The RBF kernel makes it nonlinear. A decision tree recursively partitions feature space by splitting the data at different points. The tradeoff is that the decision tree is more easy to interpret because it forms an easily readable tree structure of rules, but it yielded more false positives/negatives, while the SVC is less easier to interpret, but more accurately split the data in this dataset.**

4. Discuss how tuning hyperparameters such as regularization parameter C and kernel coefficient gamma in an SVC impacts the trade-off between bias, variance, and model generalization on unseen medical test data.

**C controls regularization, so higher values are more prone to overfitting, leading to high variance, and lower values may underfit, leading to high bias. Gamma controls the RBF decision boundary, and it works similar to C, with higher values overfitting and lower values underfitting. I used GridSearchCV to test which C and gamma values would have the best balance between bias and variance.**
