3.) Why is it critical to fit the PCA transformer *strictly* on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?

**It is critical to fit the PCA transformer strictly on the training split to prevent data leakage. If PCA is fitted on the full dataset before training downstream classifiers, it would learn from the training and test data before the LogisticRegression classifier is trained. This would lead to the LogisticRegression classifier performing almost perfectly on only this particular dataset, but it would perform poorly on new, unseen data. In order to prevent that leakage, PCA shouldn't be trained on any test data.**
