3. Explain how the Scikit-learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing steps.

**ColumnTransformer makes it easier to work with a mix of data types without having to slice the DataFrame into separate DataFrames for categorical and numerical data.
